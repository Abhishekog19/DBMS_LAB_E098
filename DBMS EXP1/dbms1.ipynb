```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="pass@123",
    database="abcd"
)

print("Connected successfully!")

cursor = conn.cursor()
cursor.execute("insert into student1(nam,sapid,mobilenumber,email) values('abcd',1234,8355897322,'hello@gmail.com')")
cursor.execute("SELECT * from student1")

rows = cursor.fetchall()

for row in rows:
    print(row)
cursor.close()
```
